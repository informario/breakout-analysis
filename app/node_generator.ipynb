```python
from pathlib import Path

from app.node_generator import generate_csv


def main() -> None:
    constraints_path = Path.cwd() / "database" / "constrains.json"
    constraints_path_free = Path.cwd() / "database" / "constrains_free.json"
    output_path = Path.cwd() / "generated_ports.csv"
    output_path_free = Path.cwd() / "generated_ports_free.csv"
    rows = 1000
    seed = 48678
    generate_csv(
        constraints_path=constraints_path,
        output_path=output_path,
        rows=rows,
        seed=seed,
    )
    generate_csv(
        constraints_path=constraints_path_free,
        output_path=output_path_free,
        rows=rows,
        seed=seed,
    )
    print(f"Wrote {output_path.resolve()}")
    print(f"Wrote {output_path_free.resolve()}")


main()
```
